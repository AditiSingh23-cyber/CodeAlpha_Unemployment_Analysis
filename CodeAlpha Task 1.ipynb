# Unemployment Rate Analysis (Raw Data)

```python
# Strip column names for raw analysis (without full cleaning)
df_area.columns = df_area.columns.str.strip()
```

```python
df_2020.columns = df_2020.columns.str.strip()
```

```python
# Overall distribution of unemployment rates across both datasets
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(df_area['Estimated Unemployment Rate (%)'].dropna(), bins=30, color='steelblue', edgecolor='white')
axes[0].set_title('Distribution of Unemployment Rate'
'(Unemployment in India Dataset)')
axes[0].set_xlabel('Unemployment Rate (%)')
axes[0].set_ylabel('Frequency')

axes[1].hist(df_2020['Estimated Unemployment Rate (%)'].dropna(), bins=30, color='darkorange', edgecolor='white')
axes[1].set_title('Distribution of Unemployment Rate'
'(Unemployment Rate upto 11/2020 Dataset)')
axes[1].set_xlabel('Unemployment Rate (%)')
axes[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
# Average unemployment rate per region (df_area - Rural/Urban dataset)
region_avg_area = df_area.groupby('Region')['Estimated Unemployment Rate (%)'].mean().sort_values(ascending=False)

plt.figure(figsize=(14, 6))
plt.bar(region_avg_area.index, region_avg_area.values, color='steelblue')
plt.title('Average Unemployment Rate by Region (Unemployment in India Dataset)')
plt.xlabel('Region')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# Rural vs Urban unemployment rate (df_area)
area_avg_raw = df_area.groupby('Area')['Estimated Unemployment Rate (%)'].mean()
print('Average Unemployment Rate by Area (Rural/Urban):')
print(area_avg_raw.round(2))

plt.figure(figsize=(6, 5))
plt.bar(area_avg_raw.index, area_avg_raw.values, color=['green', 'red'])
plt.title('Average Unemployment Rate: Rural vs Urban')
plt.xlabel('Area')
plt.ylabel('Average Unemployment Rate (%)')
plt.tight_layout()
plt.show()
```

```python
# Average unemployment rate per region (df_2020 - COVID period dataset)
region_avg_2020 = df_2020.groupby('Region')['Estimated Unemployment Rate (%)'].mean().sort_values(ascending=False)

plt.figure(figsize=(14, 6))
plt.bar(region_avg_2020.index, region_avg_2020.values, color='darkorange')
plt.title('Average Unemployment Rate by Region (Unemployment Rate upto 11/2020)')
plt.xlabel('Region')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# Unemployment rate over time (df_2020)
df_2020_temp = df_2020.copy()
df_2020_temp['Date'] = pd.to_datetime(df_2020_temp['Date'].astype(str).str.strip(), dayfirst=True, errors='coerce')
monthly_raw = df_2020_temp.groupby('Date')['Estimated Unemployment Rate (%)'].mean().reset_index()

plt.figure(figsize=(14, 5))
plt.plot(monthly_raw['Date'], monthly_raw['Estimated Unemployment Rate (%)'], marker='o', linewidth=2.5, color='darkorange')
plt.title('Monthly Average Unemployment Rate in India (2020)')
plt.xlabel('Month')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#  Clean Both Datasets Separately
# Clean Rural/Urban dataset
df_area.columns = df_area.columns.str.strip()

for col in df_area.select_dtypes(include='object').columns:
    df_area[col] = df_area[col].astype(str).str.strip()

df_area = df_area.replace({'': np.nan, 'nan': np.nan})
df_area = df_area.dropna(how='all')

df_area = df_area.dropna(subset=[ 'Region','Date','Estimated Unemployment Rate (%)', 'Area'])
```

```python
# Change the date formate

df_area['Date'] = pd.to_datetime(
    df_area['Date'].astype(str).str.strip(),
    dayfirst=True,
    errors='coerce')
```

```python
# Rename the columns name
df_area.rename(columns={
    'Estimated Unemployment Rate (%)': 'Unemployment_Rate',
    'Estimated Employed': 'Employed',
    'Estimated Labour Participation Rate (%)': 'Labour_Participation_Rate'
}, inplace=True)
```

```python
# Clean 2020 state-level dataset
df_2020.columns = df_2020.columns.str.strip()

for col in df_2020.select_dtypes(include='object').columns:
    df_2020[col] = df_2020[col].astype(str).str.strip()

df_2020.rename(columns={
    'Estimated Unemployment Rate (%)': 'Unemployment_Rate',
    'Estimated Employed': 'Employed',
    'Estimated Labour Participation Rate (%)': 'Labour_Participation_Rate',
    'Region.1': 'Zone'
}, inplace=True)
```

```python
# Change the date format
df_2020['Date'] = pd.to_datetime(
    df_2020['Date'].astype(str).str.strip(),
    dayfirst=True,
    errors='coerce')
```

```python

df_2020 = df_2020.dropna(subset=['Region', 'Date', 'Unemployment_Rate'])
```

```python
# Replace COVID Analysis Cell
before_covid = df_2020[df_2020['Date'] < '2020-03-01']

lockdown = df_2020[
    (df_2020['Date'] >= '2020-04-01') &
    (df_2020['Date'] <= '2020-05-31')
]

after_lockdown = df_2020[df_2020['Date'] >= '2020-07-01']

avg_before = before_covid['Unemployment_Rate'].mean()
avg_lockdown = lockdown['Unemployment_Rate'].mean()
avg_after = after_lockdown['Unemployment_Rate'].mean()

print(f"Before COVID average unemployment: {avg_before:.2f}%")
print(f"Lockdown average unemployment: {avg_lockdown:.2f}%")
print(f"After lockdown average unemployment: {avg_after:.2f}%")
print(f"Increase during lockdown: {avg_lockdown - avg_before:.2f} percentage points")
```

```python
#Monthly Trend
monthly_avg = df_2020.groupby('Date')['Unemployment_Rate'].mean().reset_index()

plt.figure(figsize=(14, 5))
plt.plot(
    monthly_avg['Date'],
    monthly_avg['Unemployment_Rate'],
    marker='o',
    linewidth=2.5,
    color='red'
)

plt.axvspan(
    pd.Timestamp('2020-04-01'),
    pd.Timestamp('2020-05-31'),
    alpha=0.2,
    color='orange',
    label='COVID Lockdown Peak'
)

plt.title('Monthly Unemployment Rate in India During 2020')
plt.xlabel('Month')
plt.ylabel('Unemployment Rate (%)')
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Rural vs Urban
area_avg = df_area.groupby('Area')['Unemployment_Rate'].mean()
print(area_avg)

area_trend = df_area.groupby(['Date', 'Area'])['Unemployment_Rate'].mean().reset_index()

plt.figure(figsize=(14, 5))

sns.lineplot(
    data=area_trend,
    x='Date',
    y='Unemployment_Rate',
    hue='Area',
    marker='o'
)

plt.axvspan(
    pd.Timestamp('2020-04-01'),
    pd.Timestamp('2020-05-31'),
    alpha=0.2,
    color='orange',
    label='COVID Lockdown Peak'
)

plt.title('Urban vs Rural Unemployment Rate')
plt.xlabel('Date')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# State-wise Average Unemployment Rate
state_avg = df_2020.groupby('Region')['Unemployment_Rate'].mean().sort_values(ascending=False).reset_index()

plt.figure(figsize=(14, 7))
plt.bar(state_avg['Region'], state_avg['Unemployment_Rate'], color='steelblue')
plt.title('Average Unemployment Rate by State (2020)')
plt.xlabel('State')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
monthly_avg = df_2020.groupby('Date')['Unemployment_Rate'].mean().reset_index()
peak_month = monthly_avg.loc[monthly_avg['Unemployment_Rate'].idxmax()]

most_affected = df_2020.groupby('Region')['Unemployment_Rate'].mean().idxmax()
least_affected = df_2020.groupby('Region')['Unemployment_Rate'].mean().idxmin()

print("=" * 55)
print("UNEMPLOYMENT ANALYSIS - KEY INSIGHTS")
print("=" * 55)

print(f"Average unemployment before COVID: {avg_before:.2f}%")
print(f"Average unemployment during lockdown: {avg_lockdown:.2f}%")
print(f"Average unemployment after lockdown: {avg_after:.2f}%")
print(f"COVID lockdown spike: +{avg_lockdown - avg_before:.2f} percentage points")

print(f"\nPeak unemployment month: {peak_month['Date'].strftime('%B %Y')}")
print(f"Peak monthly unemployment rate: {peak_month['Unemployment_Rate']:.2f}%")

print(f"\nMost affected region: {most_affected}")
print(f"Least affected region: {least_affected}")

print("\nUrban vs Rural average unemployment:")
print(df_area.groupby('Area')['Unemployment_Rate'].mean().round(2))

print("=" * 55)
```

## Conclusion

This project analyzed India's unemployment rate data using two datasets — one covering rural and urban unemployment trends across Indian states, and another focusing on the period leading up to and during the COVID-19 pandemic in 2020.

---

### Key Findings

#### 1. Overall Unemployment Landscape
- The unemployment rate across India showed significant variation across states and regions.
- Urban areas consistently recorded **higher unemployment (~13.17%)** compared to rural areas **(~10.32%)**, with a gap of approximately **2.84 percentage points**.
- This disparity reflects the structural difference in the nature of employment — rural economies rely more on agriculture which absorbs labor more continuously, while urban employment is more sensitive to economic shocks.

#### 2. COVID-19 Impact
- The COVID-19 pandemic and the subsequent nationwide lockdown had a **dramatic and immediate** impact on unemployment.
- Average unemployment **before COVID (Jan–Feb 2020):** ~9.23%
- Average unemployment **during lockdown (Apr–May 2020):** ~22.75%
- This represents a spike of **+13.52 percentage points** in just a few months — nearly **2.5× the pre-COVID level**.
- **May 2020** recorded the peak monthly unemployment rate of **23.24%**, the highest in the dataset.

#### 3. Recovery
- After the lockdown eased (July 2020 onwards), unemployment fell back to **~9.22%**, almost identical to pre-COVID levels, suggesting a **strong but swift recovery** once restrictions were lifted.
- This V-shaped recovery indicates that the unemployment shock was largely temporary and driven by lockdown restrictions rather than long-term structural changes.

#### 4. Regional Disparities
- **Haryana** was the most affected state, recording the highest average unemployment rate during 2020.
- **Meghalaya** was the least affected, likely due to its rural, agriculture-dependent economy which was less disrupted by lockdowns.
- These regional differences highlight that economic vulnerability to pandemics is not uniform — states with larger urban and industrial workforces face greater risk.

---

### Policy Implications

| Observation | Suggested Policy Action |
|---|---|
| Urban unemployment > Rural | Strengthen urban employment safety nets and skilling programs |
| Severe lockdown spike | Develop rapid-response economic relief mechanisms for future crises |
| Regional disparities | Targeted support for high-vulnerability states like Haryana during shocks |
| Swift post-lockdown recovery | Invest in economic resilience to sustain recovery momentum |

---

### Summary

The analysis clearly demonstrates that **COVID-19 caused an unprecedented short-term unemployment shock in India**, with the hardest impact felt in urban areas and industrialized states. However, India's labor market showed remarkable resilience, recovering quickly once restrictions were lifted. These insights can help policymakers design more responsive and targeted economic interventions for future crises.
